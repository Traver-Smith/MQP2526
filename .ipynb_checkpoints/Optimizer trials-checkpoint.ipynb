```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()
opt_sdae = optim.Adam(sdae.parameters(), lr=1e-3)
opt_dae  = optim.Adam(dae.parameters(),  lr=1e-3)
opt_ae   = optim.Adam(ae.parameters(),   lr=1e-3)

noise_factor = 0.81
epochs = 10
stack_k_train = 4     # SDAE training re-feeds
stack_k_eval  = 4    # SDAE default eval re-feeds

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        x = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(x, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        opt_sdae.step()

        # ---- DAE: noisy -> clean with 1 pass
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        opt_dae.step()

        # ---- AE: clean -> clean with 1 pass
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        opt_ae.step()

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
baseline_adam_sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [baseline_adam_sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import torch
import torch.nn as nn
from torch.optim import AdamW
from torchvision import datasets, transforms
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)

train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True,  num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def evaluate(model: nn.Module, loader, *, denoising: bool, re_feeds: int = 1,
             noise_factor: float = 0.81, device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)
# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)
# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()
base_lr   = 3e-4
opt_sdae  = AdamW(sdae.parameters(), lr=base_lr, betas=(0.9, 0.95))
opt_dae   = AdamW(dae.parameters(),  lr=base_lr, betas=(0.9, 0.95))
opt_ae    = AdamW(ae.parameters(),   lr=base_lr, betas=(0.9, 0.95))

noise_factor = 0.81
epochs        = 10
stack_k_train = 3     # SDAE training re-feeds
stack_k_eval  = 3     # SDAE default eval re-feeds

# ----------------------------
# Training loop (all three together)
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs  = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)
        bsz   = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        x = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(x, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)   # per-step clipping
        opt_sdae.step()

        # ---- DAE: noisy -> clean with 1 pass
        out_dae  = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)    # per-step clipping
        opt_dae.step()

        # ---- AE: clean -> clean with 1 pass
        out_ae  = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)     # per-step clipping
        opt_ae.step()

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse  = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
adamW_sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [adamW_sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)

    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize:
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math
import torch
import torch.nn as nn
import torch.optim as optim
from torch.optim import AdamW
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """ Add Gaussian noise to an input image tensor in [0,1], then clamp. """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)

train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True,  num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )

    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """ Feed 'x' through 'model' k times (re-feeds). """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module, loader, *, denoising: bool, re_feeds: int = 1,
             noise_factor: float = 0.81, device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """ Evaluate reconstruction MSE against the clean target images.

    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# --- wrapper utils (ADDED) ---
def split_param_groups(model: nn.Module):
    """Decouple weight decay for biases and 1D params (e.g., LayerNorm)."""
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if p.ndim == 1 or name.endswith(".bias"):
            no_decay.append(p)
        else:
            decay.append(p)
    return [
        {"params": decay, "weight_decay": 0.01},
        {"params": no_decay, "weight_decay": 0.0},
    ]

def build_warmup_cosine(optimizer, total_steps: int, warmup_frac: float, eta_min: float, base_lr: float):
    """Returns a callable step_lr(global_step) that sets LR on the optimizer."""
    warmup_steps = max(1, int(total_steps * warmup_frac))
    remain = max(1, total_steps - warmup_steps)

    def step_lr(step: int):
        if step <= warmup_steps:
            lr = base_lr * step / warmup_steps
        else:
            t = (step - warmup_steps) / remain
            lr = eta_min + 0.5 * (base_lr - eta_min) * (1.0 + math.cos(math.pi * t))
        for pg in optimizer.param_groups:
            pg["lr"] = lr

    return step_lr

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)
# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae = DenoisingAutoencoder().to(device)
# 3) Plain AE: trained 1 pass on clean inputs
ae = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# --- Optimizers with param groups + AdamW (CHANGED) ---
base_lr = 3e-4
opt_sdae = AdamW(split_param_groups(sdae), lr=base_lr, betas=(0.9, 0.95))
opt_dae  = AdamW(split_param_groups(dae),  lr=base_lr, betas=(0.9, 0.95))
opt_ae   = AdamW(split_param_groups(ae),   lr=base_lr, betas=(0.9, 0.95))

noise_factor = 0.81
epochs = 10
stack_k_train = 3  # SDAE training re-feeds
stack_k_eval  = 3  # SDAE default eval re-feeds

# --- LR schedules & global step (NEW) ---
total_steps = epochs * len(train_loader)
step_lr_sdae = build_warmup_cosine(opt_sdae, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_dae  = build_warmup_cosine(opt_dae,  total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_ae   = build_warmup_cosine(opt_ae,   total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
global_step = 0

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)
        bsz = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        x = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(x, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)   # NEW
        opt_sdae.step()
        step_lr_sdae(global_step)                                # NEW

        # ---- DAE: noisy -> clean with 1 pass
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)    # NEW
        opt_dae.step()
        step_lr_dae(global_step)                                 # NEW

        # ---- AE: clean -> clean with 1 pass
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)     # NEW
        opt_ae.step()
        step_lr_ae(global_step)                                  # NEW

        # accounting
        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item() * bsz
        train_sum["ae"]   += loss_ae.item() * bsz
        train_cnt += bsz

        global_step += 1  # NEW (increment once per batch)

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE train_MSE={train_sum['dae']/train_cnt:.4f} test_MSE={dae_test:.4f} | "
        f"AE train_MSE={train_sum['ae']/train_cnt:.4f} test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module, loader, ks: List[int], noise_factor: float,
                     device: torch.device, criterion = nn.MSELoss()) -> Dict[int, float]:
    """ Compute test MSE for each k re-feeds on the *same* test set, denoising mode. Returns dict {k: mse}. """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k, noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse  = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
AdamW_per_step_clipping_cosine_warmupsdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [AdamW_per_step_clipping_cosine_warmupsdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':', label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-bfeed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)

    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math  # <-- added
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Optim helpers (added)
# ----------------------------
def split_param_groups(model: nn.Module, weight_decay: float = 0.01):
    """
    Typical AdamW param grouping:
    - decay: all params with ndim >= 2
    - no_decay: biases and LayerNorm/BatchNorm/any 1D params
    """
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if p.ndim >= 2:
            decay.append(p)
        else:
            no_decay.append(p)
    return [
        {"params": decay, "weight_decay": weight_decay},
        {"params": no_decay, "weight_decay": 0.0},
    ]

def build_warmup_cosine(optimizer, total_steps: int, warmup_frac: float = 0.05, eta_min: float = 3e-5, base_lr: float = 3e-4):
    """
    Returns a closure step_lr(global_step) that sets LR per step.
    - Linear warmup to base_lr for first warmup_steps
    - Cosine decay to eta_min for the rest
    """
    warmup_steps = max(1, int(warmup_frac * total_steps))

    def set_lr(lr):
        for pg in optimizer.param_groups:
            # respect per-group base lr if set
            pg_lr = pg.get("initial_lr", pg.get("lr", base_lr))
            scale = lr / base_lr
            pg["lr"] = pg_lr * scale

    # store initial lr for each group
    for pg in optimizer.param_groups:
        pg.setdefault("initial_lr", pg.get("lr", base_lr))

    def lr_fn(step):
        if step <= warmup_steps:
            return base_lr * step / warmup_steps
        # cosine from base_lr -> eta_min
        t = (step - warmup_steps) / max(1, (total_steps - warmup_steps))
        return eta_min + 0.5 * (base_lr - eta_min) * (1 + math.cos(math.pi * t))

    def step_lr(global_step: int):
        lr = lr_fn(min(global_step, total_steps))
        set_lr(lr)

    return step_lr

# ----------------------------
# SAM wrapper (added)
# ----------------------------
class SAM:
    def __init__(self, model, base_optimizer, rho=0.05, eps=1e-12):
        self.model = model
        self.optimizer = base_optimizer
        self.rho = rho
        self.eps = eps
        self.state = {}

    @torch.no_grad()
    def _grad_norm(self):
        norms = [p.grad.detach().norm(2) for p in self.model.parameters() if p.grad is not None]
        if not norms: return torch.tensor(0., device=next(self.model.parameters()).device)
        return torch.norm(torch.stack(norms), 2)

    @torch.no_grad()
    def first_step(self):
        scale = self.rho / (self._grad_norm() + self.eps)
        self.state.clear()
        for p in self.model.parameters():
            if p.grad is None: continue
            e = p.grad * scale
            p.add_(e)
            self.state[p] = e

    @torch.no_grad()
    def second_step(self):
        for p, e in self.state.items():
            p.sub_(e)
        self.optimizer.step()
        self.optimizer.zero_grad(set_to_none=True)
        self.state.clear()

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# --- build optimizer & schedules (updated) ---
base_lr = 3e-4
epochs = 10

# Use param groups for AdamW
opt_sdae_base = optim.AdamW(split_param_groups(sdae), lr=base_lr, betas=(0.9, 0.95))
opt_dae       = optim.AdamW(split_param_groups(dae),  lr=base_lr, betas=(0.9, 0.95))
opt_ae        = optim.AdamW(split_param_groups(ae),   lr=base_lr, betas=(0.9, 0.95))

# SAM only for SDAE (most sensitive to sharp minima)
sam_sdae = SAM(sdae, opt_sdae_base, rho=0.04)

total_steps = epochs * len(train_loader)
step_lr_sdae = build_warmup_cosine(opt_sdae_base, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_dae  = build_warmup_cosine(opt_dae,       total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_ae   = build_warmup_cosine(opt_ae,        total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

# global step counters
global_step_sdae = 0
global_step_dae  = 0
global_step_ae   = 0

noise_factor = 0.81
stack_k_train = 3     # SDAE training re-feeds
stack_k_eval  = 3     # SDAE default eval re-feeds

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE (with SAM): noisy -> clean with k re-feeds
        out_sdae = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae_base.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        sam_sdae.first_step()  # w <- w + e

        # second forward/backward at perturbed weights
        out_sdae_pert = forward_k(sdae, noisy, stack_k_train)
        loss_sdae_pert = criterion(out_sdae_pert, imgs)
        opt_sdae_base.zero_grad(set_to_none=True)
        loss_sdae_pert.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        sam_sdae.second_step()  # restore, step, zero_grad

        global_step_sdae += 1
        step_lr_sdae(global_step_sdae)

        # ---- DAE: noisy -> clean with 1 pass (AdamW + sched + clip)
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        global_step_dae += 1
        step_lr_dae(global_step_dae)

        # ---- AE: clean -> clean with 1 pass (AdamW + sched + clip)
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        global_step_ae += 1
        step_lr_ae(global_step_ae)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
AdamW_samWrapper_sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [AdamW_samWrapper_sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math  # <-- added
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Optim helpers (added)
# ----------------------------
def split_param_groups(model: nn.Module, weight_decay: float = 0.01):
    """
    Typical AdamW param grouping:
    - decay: all params with ndim >= 2
    - no_decay: biases and LayerNorm/BatchNorm/any 1D params
    """
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if p.ndim >= 2:
            decay.append(p)
        else:
            no_decay.append(p)
    return [
        {"params": decay, "weight_decay": weight_decay},
        {"params": no_decay, "weight_decay": 0.0},
    ]

def build_warmup_cosine(optimizer, total_steps: int, warmup_frac: float = 0.05, eta_min: float = 3e-5, base_lr: float = 3e-4):
    """
    Returns a closure step_lr(global_step) that sets LR per step.
    - Linear warmup to base_lr for first warmup_steps
    - Cosine decay to eta_min for the rest
    """
    warmup_steps = max(1, int(warmup_frac * total_steps))

    def set_lr(lr):
        for pg in optimizer.param_groups:
            # respect per-group base lr if set
            pg_lr = pg.get("initial_lr", pg.get("lr", base_lr))
            scale = lr / base_lr
            pg["lr"] = pg_lr * scale

    # store initial lr for each group
    for pg in optimizer.param_groups:
        pg.setdefault("initial_lr", pg.get("lr", base_lr))

    def lr_fn(step):
        if step <= warmup_steps:
            return base_lr * step / warmup_steps
        # cosine from base_lr -> eta_min
        t = (step - warmup_steps) / max(1, (total_steps - warmup_steps))
        return eta_min + 0.5 * (base_lr - eta_min) * (1 + math.cos(math.pi * t))

    def step_lr(global_step: int):
        lr = lr_fn(min(global_step, total_steps))
        set_lr(lr)

    return step_lr

# ----------------------------
# SAM wrapper (added)
# ----------------------------
class SAM:
    def __init__(self, model, base_optimizer, rho=0.05, eps=1e-12):
        self.model = model
        self.optimizer = base_optimizer
        self.rho = rho
        self.eps = eps
        self.state = {}

    @torch.no_grad()
    def _grad_norm(self):
        norms = [p.grad.detach().norm(2) for p in self.model.parameters() if p.grad is not None]
        if not norms: return torch.tensor(0., device=next(self.model.parameters()).device)
        return torch.norm(torch.stack(norms), 2)

    @torch.no_grad()
    def first_step(self):
        scale = self.rho / (self._grad_norm() + self.eps)
        self.state.clear()
        for p in self.model.parameters():
            if p.grad is None: continue
            e = p.grad * scale
            p.add_(e)
            self.state[p] = e

    @torch.no_grad()
    def second_step(self):
        for p, e in self.state.items():
            p.sub_(e)
        self.optimizer.step()
        self.optimizer.zero_grad(set_to_none=True)
        self.state.clear()

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# --- build optimizer & schedules (updated) ---
base_lr = 3e-4
epochs = 30

# Use param groups for AdamW
opt_sdae_base = optim.AdamW(split_param_groups(sdae), lr=base_lr, betas=(0.9, 0.95))
opt_dae       = optim.AdamW(split_param_groups(dae),  lr=base_lr, betas=(0.9, 0.95))
opt_ae        = optim.AdamW(split_param_groups(ae),   lr=base_lr, betas=(0.9, 0.95))

# SAM only for SDAE (most sensitive to sharp minima)
sam_sdae = SAM(sdae, opt_sdae_base, rho=0.04)

total_steps = epochs * len(train_loader)
step_lr_sdae = build_warmup_cosine(opt_sdae_base, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_dae  = build_warmup_cosine(opt_dae,       total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_ae   = build_warmup_cosine(opt_ae,        total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

# global step counters
global_step_sdae = 0
global_step_dae  = 0
global_step_ae   = 0

noise_factor = 0.81
stack_k_train = 3     # SDAE training re-feeds
stack_k_eval  = 3     # SDAE default eval re-feeds

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE (with SAM): noisy -> clean with k re-feeds
        out_sdae = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae_base.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        sam_sdae.first_step()  # w <- w + e

        # second forward/backward at perturbed weights
        out_sdae_pert = forward_k(sdae, noisy, stack_k_train)
        loss_sdae_pert = criterion(out_sdae_pert, imgs)
        opt_sdae_base.zero_grad(set_to_none=True)
        loss_sdae_pert.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        sam_sdae.second_step()  # restore, step, zero_grad

        global_step_sdae += 1
        step_lr_sdae(global_step_sdae)

        # ---- DAE: noisy -> clean with 1 pass (AdamW + sched + clip)
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        global_step_dae += 1
        step_lr_dae(global_step_dae)

        # ---- AE: clean -> clean with 1 pass (AdamW + sched + clip)
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        global_step_ae += 1
        step_lr_ae(global_step_ae)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math  # <-- added
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Optim helpers (added)
# ----------------------------
def split_param_groups(model: nn.Module, weight_decay: float = 0.01):
    """
    Typical AdamW param grouping:
    - decay: all params with ndim >= 2
    - no_decay: biases and LayerNorm/BatchNorm/any 1D params
    """
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if p.ndim >= 2:
            decay.append(p)
        else:
            no_decay.append(p)
    return [
        {"params": decay, "weight_decay": weight_decay},
        {"params": no_decay, "weight_decay": 0.0},
    ]

def build_warmup_cosine(optimizer, total_steps: int, warmup_frac: float = 0.05, eta_min: float = 3e-5, base_lr: float = 3e-4):
    """
    Returns a closure step_lr(global_step) that sets LR per step.
    - Linear warmup to base_lr for first warmup_steps
    - Cosine decay to eta_min for the rest
    """
    warmup_steps = max(1, int(warmup_frac * total_steps))

    def set_lr(lr):
        for pg in optimizer.param_groups:
            # respect per-group base lr if set
            pg_lr = pg.get("initial_lr", pg.get("lr", base_lr))
            scale = lr / base_lr
            pg["lr"] = pg_lr * scale

    # store initial lr for each group
    for pg in optimizer.param_groups:
        pg.setdefault("initial_lr", pg.get("lr", base_lr))

    def lr_fn(step):
        if step <= warmup_steps:
            return base_lr * step / warmup_steps
        # cosine from base_lr -> eta_min
        t = (step - warmup_steps) / max(1, (total_steps - warmup_steps))
        return eta_min + 0.5 * (base_lr - eta_min) * (1 + math.cos(math.pi * t))

    def step_lr(global_step: int):
        lr = lr_fn(min(global_step, total_steps))
        set_lr(lr)

    return step_lr

# ----------------------------
# Lookahead wrapper (added)
# ----------------------------
class Lookahead(torch.optim.Optimizer):
    def __init__(self, base_optimizer, k=5, alpha=0.5):
        self.optimizer = base_optimizer
        self.k = k
        self.alpha = alpha
        self.step_counter = 0
        # snapshot slow params
        self.slow_params = [p.clone().detach() for group in base_optimizer.param_groups for p in group['params']]
        for p in self.slow_params:
            p.requires_grad_(False)

    @property
    def param_groups(self):
        # expose base param_groups so schedulers can see/update LR
        return self.optimizer.param_groups

    def zero_grad(self, **kw): 
        return self.optimizer.zero_grad(**kw)

    @torch.no_grad()
    def step(self, closure=None):
        loss = self.optimizer.step(closure)
        self.step_counter += 1
        if self.step_counter % self.k == 0:
            idx = 0
            for group in self.optimizer.param_groups:
                for p in group['params']:
                    if p is None:
                        continue
                    s = self.slow_params[idx]
                    s.add_(p.data - s, alpha=self.alpha)  # slow = slow + alpha*(fast - slow)
                    p.data.copy_(s)
                    idx += 1
        return loss

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)
# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)
# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# --- build optimizer & schedules (updated to Lookahead over AdamW) ---
base_lr = 3e-4
epochs = 10

# base AdamW with param groups
opt_sdae_base = optim.AdamW(split_param_groups(sdae), lr=base_lr, betas=(0.9, 0.95))
opt_dae_base  = optim.AdamW(split_param_groups(dae),  lr=base_lr, betas=(0.9, 0.95))
opt_ae_base   = optim.AdamW(split_param_groups(ae),   lr=base_lr, betas=(0.9, 0.95))

# Lookahead wrappers
opt_sdae = Lookahead(opt_sdae_base, k=5, alpha=0.5)
opt_dae  = Lookahead(opt_dae_base,  k=5, alpha=0.5)
opt_ae   = Lookahead(opt_ae_base,   k=5, alpha=0.5)

total_steps  = epochs * len(train_loader)
step_lr_sdae = build_warmup_cosine(opt_sdae_base, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_dae  = build_warmup_cosine(opt_dae_base,  total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_ae   = build_warmup_cosine(opt_ae_base,   total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

# per-model global steps (so schedulers stay aligned)
global_step_sdae = 0
global_step_dae  = 0
global_step_ae   = 0

noise_factor   = 0.81
stack_k_train  = 3     # SDAE training re-feeds
stack_k_eval   = 3     # SDAE default eval re-feeds

# ----------------------------
# Training loop (all three together)
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs  = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)
        bsz   = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds (Lookahead+AdamW, warmup+cosine, per-step clip)
        out_sdae  = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        opt_sdae.step()
        global_step_sdae += 1
        step_lr_sdae(global_step_sdae)

        # ---- DAE: noisy -> clean with 1 pass
        out_dae  = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        global_step_dae += 1
        step_lr_dae(global_step_dae)

        # ---- AE: clean -> clean with 1 pass
        out_ae  = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        global_step_ae += 1
        step_lr_ae(global_step_ae)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
AdamW_Lookahead_wrapper_sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [AdamW_Lookahead_wrapper_sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math  # <-- added
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torchvision.utils import make_grid, save_image
import numpy as np
import matplotlib.pyplot as plt
from torchvision.utils import make_grid
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Optim helpers (added)
# ----------------------------
def split_param_groups(model: nn.Module, weight_decay: float = 0.01):
    """
    Typical AdamW param grouping:
    - decay: all params with ndim >= 2
    - no_decay: biases and LayerNorm/BatchNorm/any 1D params
    """
    decay, no_decay = [], []
    for name, p in model.named_parameters():
        if not p.requires_grad:
            continue
        if p.ndim >= 2:
            decay.append(p)
        else:
            no_decay.append(p)
    return [
        {"params": decay, "weight_decay": weight_decay},
        {"params": no_decay, "weight_decay": 0.0},
    ]

def build_warmup_cosine(optimizer, total_steps: int, warmup_frac: float = 0.05, eta_min: float = 3e-5, base_lr: float = 3e-4):
    """
    Returns a closure step_lr(global_step) that sets LR per step.
    - Linear warmup to base_lr for first warmup_steps
    - Cosine decay to eta_min for the rest
    """
    warmup_steps = max(1, int(warmup_frac * total_steps))

    def set_lr(lr):
        for pg in optimizer.param_groups:
            # respect per-group base lr if set
            pg_lr = pg.get("initial_lr", pg.get("lr", base_lr))
            scale = lr / base_lr
            pg["lr"] = pg_lr * scale

    # store initial lr for each group
    for pg in optimizer.param_groups:
        pg.setdefault("initial_lr", pg.get("lr", base_lr))

    def lr_fn(step):
        if step <= warmup_steps:
            return base_lr * step / warmup_steps
        # cosine from base_lr -> eta_min
        t = (step - warmup_steps) / max(1, (total_steps - warmup_steps))
        return eta_min + 0.5 * (base_lr - eta_min) * (1 + math.cos(math.pi * t))

    def step_lr(global_step: int):
        lr = lr_fn(min(global_step, total_steps))
        set_lr(lr)

    return step_lr

# ----------------------------
# Lookahead wrapper (added)
# ----------------------------
class Lookahead(torch.optim.Optimizer):
    def __init__(self, base_optimizer, k=5, alpha=0.5):
        self.optimizer = base_optimizer
        self.k = k
        self.alpha = alpha
        self.step_counter = 0
        # snapshot slow params
        self.slow_params = [p.clone().detach() for group in base_optimizer.param_groups for p in group['params']]
        for p in self.slow_params:
            p.requires_grad_(False)

    @property
    def param_groups(self):
        # expose base param_groups so schedulers can see/update LR
        return self.optimizer.param_groups

    def zero_grad(self, **kw): 
        return self.optimizer.zero_grad(**kw)

    @torch.no_grad()
    def step(self, closure=None):
        loss = self.optimizer.step(closure)
        self.step_counter += 1
        if self.step_counter % self.k == 0:
            idx = 0
            for group in self.optimizer.param_groups:
                for p in group['params']:
                    if p is None:
                        continue
                    s = self.slow_params[idx]
                    s.add_(p.data - s, alpha=self.alpha)  # slow = slow + alpha*(fast - slow)
                    p.data.copy_(s)
                    idx += 1
        return loss

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)
# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)
# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# --- build optimizer & schedules (updated to Lookahead over AdamW) ---
base_lr = 3e-4
epochs = 22

# base AdamW with param groups
opt_sdae_base = optim.AdamW(split_param_groups(sdae), lr=base_lr, betas=(0.9, 0.95))
opt_dae_base  = optim.AdamW(split_param_groups(dae),  lr=base_lr, betas=(0.9, 0.95))
opt_ae_base   = optim.AdamW(split_param_groups(ae),   lr=base_lr, betas=(0.9, 0.95))

# Lookahead wrappers
opt_sdae = Lookahead(opt_sdae_base, k=5, alpha=0.5)
opt_dae  = Lookahead(opt_dae_base,  k=5, alpha=0.5)
opt_ae   = Lookahead(opt_ae_base,   k=5, alpha=0.5)

total_steps  = epochs * len(train_loader)
step_lr_sdae = build_warmup_cosine(opt_sdae_base, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_dae  = build_warmup_cosine(opt_dae_base,  total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
step_lr_ae   = build_warmup_cosine(opt_ae_base,   total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

# per-model global steps (so schedulers stay aligned)
global_step_sdae = 0
global_step_dae  = 0
global_step_ae   = 0

noise_factor   = 0.81
stack_k_train  = 3     # SDAE training re-feeds
stack_k_eval   = 3     # SDAE default eval re-feeds

# ----------------------------
# Training loop (all three together)
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs  = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)
        bsz   = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds (Lookahead+AdamW, warmup+cosine, per-step clip)
        out_sdae  = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        opt_sdae.step()
        global_step_sdae += 1
        step_lr_sdae(global_step_sdae)

        # ---- DAE: noisy -> clean with 1 pass
        out_dae  = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        global_step_dae += 1
        step_lr_dae(global_step_dae)

        # ---- AE: clean -> clean with 1 pass
        out_ae  = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        global_step_ae += 1
        step_lr_ae(global_step_ae)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,             noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, color='gray', linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  color='black', linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# ----------------------------
# Adan + Warmup-Cosine + grad clipping
# ----------------------------
from adan_pytorch import Adan#, build_warmup_cosine

base_lr       = 3e-4
weight_decay  = 1e-4
betas         = (0.98, 0.92, 0.99)
epochs        = 10
stack_k_train = 3     # SDAE training re-feeds
stack_k_eval  = 3     # SDAE default eval re-feeds
noise_factor  = 0.81

opt_sdae = Adan(sdae.parameters(), lr=base_lr, weight_decay=weight_decay, betas=betas)
opt_dae  = Adan(dae.parameters(),  lr=base_lr, weight_decay=weight_decay, betas=betas)
opt_ae   = Adan(ae.parameters(),   lr=base_lr, weight_decay=weight_decay, betas=betas)

total_steps = epochs * len(train_loader)
sched_sdae = build_warmup_cosine(opt_sdae, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
sched_dae  = build_warmup_cosine(opt_dae,  total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
sched_ae   = build_warmup_cosine(opt_ae,   total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

global_step = 0

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        out_sdae = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        opt_sdae.step()

        global_step += 1
        sched_sdae(global_step)

        # ---- DAE: noisy -> clean with 1 pass
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        sched_dae(global_step)

        # ---- AE: clean -> clean with 1 pass
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        sched_ae(global_step)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
Adan_warmup_cosine_grad_clipping = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [Adan_warmup_cosine_grad_clipping[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize some examples after training:
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import torch
import torch.nn as nn
from torchvision import datasets, transforms
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# ----------------------------
# Adan + Warmup-Cosine + grad clipping
# ----------------------------
from adan_pytorch import Adan#, build_warmup_cosine

base_lr       = 3e-4
weight_decay  = 1e-4
betas         = (0.98, 0.92, 0.99)
epochs        = 20
stack_k_train = 3     # SDAE training re-feeds
stack_k_eval  = 3     # SDAE default eval re-feeds
noise_factor  = 0.81

opt_sdae = Adan(sdae.parameters(), lr=base_lr, weight_decay=weight_decay, betas=betas)
opt_dae  = Adan(dae.parameters(),  lr=base_lr, weight_decay=weight_decay, betas=betas)
opt_ae   = Adan(ae.parameters(),   lr=base_lr, weight_decay=weight_decay, betas=betas)

total_steps = epochs * len(train_loader)
sched_sdae = build_warmup_cosine(opt_sdae, total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
sched_dae  = build_warmup_cosine(opt_dae,  total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)
sched_ae   = build_warmup_cosine(opt_ae,   total_steps, warmup_frac=0.05, eta_min=3e-5, base_lr=base_lr)

global_step = 0

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        out_sdae = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        opt_sdae.step()

        global_step += 1
        sched_sdae(global_step)

        # ---- DAE: noisy -> clean with 1 pass
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        sched_dae(global_step)

        # ---- AE: clean -> clean with 1 pass
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        sched_ae(global_step)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
sdae_curve = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [sdae_curve[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize some examples after training:
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```

```python
import math
import torch
import torch.nn as nn
from torch.optim import SGD
from torchvision import datasets, transforms
from torchvision.utils import make_grid
import matplotlib.pyplot as plt
from typing import Dict, List

# ----------------------------
# Noise
# ----------------------------
def add_noise(img, noise_factor=0.81):
    """
    Add Gaussian noise to an input image tensor in [0,1], then clamp.
    """
    noisy = img + noise_factor * torch.randn_like(img)
    return torch.clamp(noisy, 0., 1.)

# ----------------------------
# Data
# ----------------------------
transform = transforms.ToTensor()
train_data = datasets.MNIST(root='./data', train=True, transform=transform, download=True)
test_data  = datasets.MNIST(root='./data', train=False, transform=transform, download=True)
train_loader = torch.utils.data.DataLoader(train_data, batch_size=128, shuffle=True, num_workers=0, pin_memory=True)
test_loader  = torch.utils.data.DataLoader(test_data,  batch_size=256, shuffle=False, num_workers=0, pin_memory=True)

# ----------------------------
# Model
# ----------------------------
class DenoisingAutoencoder(nn.Module):
    def __init__(self):
        super().__init__()
        self.encoder = nn.Sequential(
            nn.Flatten(),
            nn.Linear(784, 512), nn.ReLU(),
            nn.Linear(512, 128), nn.ReLU(),
            nn.Linear(128, 64)
        )
        self.decoder = nn.Sequential(
            nn.Linear(64, 128), nn.ReLU(),
            nn.Linear(128, 512), nn.ReLU(),
            nn.Linear(512, 784), nn.Sigmoid(),
            nn.Unflatten(1, (1, 28, 28))
        )
    def forward(self, x):
        z = self.encoder(x)
        return self.decoder(z)

# ----------------------------
# Utilities
# ----------------------------
def forward_k(model: nn.Module, x: torch.Tensor, k: int) -> torch.Tensor:
    """
    Feed 'x' through 'model' k times (re-feeds).
    """
    for _ in range(k):
        x = model(x)
    return x

@torch.no_grad()
def batch_mse(a, b):
    return ((a - b) ** 2).mean().item()

def evaluate(model: nn.Module,
             loader,
             *,
             denoising: bool,
             re_feeds: int = 1,
             noise_factor: float = 0.81,
             device: torch.device = torch.device("cpu"),
             criterion = nn.MSELoss()) -> float:
    """
    Evaluate reconstruction MSE against the clean target images.
    - If denoising=True, inputs = add_noise(clean), targets = clean.
    - If denoising=False, inputs = clean, targets = clean.
    - The model output is forward_k(model, inputs, re_feeds).
    """
    model.eval()
    total, count = 0.0, 0
    with torch.no_grad():
        for imgs, _ in loader:
            imgs = imgs.to(device, non_blocking=True)
            inputs = add_noise(imgs, noise_factor) if denoising else imgs
            recon = forward_k(model, inputs, re_feeds)
            loss = criterion(recon, imgs)
            bsz = imgs.size(0)
            total += loss.item() * bsz
            count += bsz
    return total / max(count, 1)

# ----------------------------
# Warmup + Cosine scheduler builder (per-optimizer)
# ----------------------------
def build_warmup_cosine(optimizer,
                        total_steps: int,
                        warmup_frac: float = 0.05,
                        eta_min: float = 5e-4,
                        base_lr: float = 5e-3):
    warmup_steps = max(1, int(total_steps * warmup_frac))
    cosine_steps = max(1, total_steps - warmup_steps)

    def step_lr(global_step: int):
        if global_step <= warmup_steps:
            # linear warmup from eta_min -> base_lr
            lr = eta_min + (base_lr - eta_min) * (global_step / warmup_steps)
        else:
            t = (global_step - warmup_steps) / cosine_steps
            t = min(max(t, 0.0), 1.0)
            lr = eta_min + 0.5 * (base_lr - eta_min) * (1.0 + math.cos(math.pi * t))
        for g in optimizer.param_groups:
            g['lr'] = lr
        return lr

    return step_lr

# ----------------------------
# Setup 3 models
# ----------------------------
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# 1) Stacked denoising AE (SDAE): trained with k>1 re-feeds on noisy inputs
sdae = DenoisingAutoencoder().to(device)

# 2) Single-pass denoising AE (DAE): trained 1 pass on noisy inputs
dae  = DenoisingAutoencoder().to(device)

# 3) Plain AE: trained 1 pass on clean inputs
ae   = DenoisingAutoencoder().to(device)

criterion = nn.MSELoss()

# ----------------------------
# SGD + Warmup-Cosine + grad clipping
# ----------------------------
base_lr       = 5e-3   # tune 1e-3..1e-2
weight_decay  = 1e-4
momentum      = 0.9
epochs        = 10
stack_k_train = 3
stack_k_eval  = 3
noise_factor  = 0.81

opt_sdae = SGD(sdae.parameters(), lr=base_lr, momentum=momentum, nesterov=True, weight_decay=weight_decay)
opt_dae  = SGD(dae.parameters(),  lr=base_lr, momentum=momentum, nesterov=True, weight_decay=weight_decay)
opt_ae   = SGD(ae.parameters(),   lr=base_lr, momentum=momentum, nesterov=True, weight_decay=weight_decay)

total_steps = epochs * len(train_loader)
sched_sdae = build_warmup_cosine(opt_sdae, total_steps, warmup_frac=0.05, eta_min=5e-4, base_lr=base_lr)
sched_dae  = build_warmup_cosine(opt_dae,  total_steps, warmup_frac=0.05, eta_min=5e-4, base_lr=base_lr)
sched_ae   = build_warmup_cosine(opt_ae,   total_steps, warmup_frac=0.05, eta_min=5e-4, base_lr=base_lr)

global_step = 0

# ----------------------------
# Training loop
# ----------------------------
for epoch in range(1, epochs + 1):
    sdae.train(); dae.train(); ae.train()
    train_sum = {"sdae":0.0, "dae":0.0, "ae":0.0}
    train_cnt = 0

    for imgs, _ in train_loader:
        imgs = imgs.to(device, non_blocking=True)
        noisy = add_noise(imgs, noise_factor)

        bsz = imgs.size(0)

        # ---- SDAE: noisy -> clean with k re-feeds
        out_sdae = forward_k(sdae, noisy, stack_k_train)
        loss_sdae = criterion(out_sdae, imgs)
        opt_sdae.zero_grad(set_to_none=True)
        loss_sdae.backward()
        torch.nn.utils.clip_grad_norm_(sdae.parameters(), 1.0)
        opt_sdae.step()

        global_step += 1
        sched_sdae(global_step)

        # ---- DAE: noisy -> clean with 1 pass
        out_dae = dae(noisy)
        loss_dae = criterion(out_dae, imgs)
        opt_dae.zero_grad(set_to_none=True)
        loss_dae.backward()
        torch.nn.utils.clip_grad_norm_(dae.parameters(), 1.0)
        opt_dae.step()
        sched_dae(global_step)

        # ---- AE: clean -> clean with 1 pass
        out_ae = ae(imgs)
        loss_ae = criterion(out_ae, imgs)
        opt_ae.zero_grad(set_to_none=True)
        loss_ae.backward()
        torch.nn.utils.clip_grad_norm_(ae.parameters(), 1.0)
        opt_ae.step()
        sched_ae(global_step)

        train_sum["sdae"] += loss_sdae.item() * bsz
        train_sum["dae"]  += loss_dae.item()  * bsz
        train_sum["ae"]   += loss_ae.item()   * bsz
        train_cnt         += bsz

    # ---- Eval in natural regimes
    sdae_test = evaluate(sdae, test_loader, denoising=True,  re_feeds=stack_k_eval, noise_factor=noise_factor, device=device, criterion=criterion)
    dae_test  = evaluate(dae,  test_loader, denoising=True,  re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)
    ae_test   = evaluate(ae,   test_loader, denoising=False, re_feeds=1,               noise_factor=noise_factor, device=device, criterion=criterion)

    print(
        f"Epoch {epoch:02d} | "
        f"SDAE train_MSE={train_sum['sdae']/train_cnt:.4f} test_MSE={sdae_test:.4f} (k_train={stack_k_train}, k_eval={stack_k_eval}) | "
        f"DAE  train_MSE={train_sum['dae']/train_cnt:.4f}  test_MSE={dae_test:.4f} | "
        f"AE   train_MSE={train_sum['ae']/train_cnt:.4f}   test_MSE={ae_test:.4f}"
    )

# ----------------------------
# Iterative re-feed sweep for SDAE (k=1..100), with baselines
# ----------------------------
sdae.eval(); dae.eval(); ae.eval()

@torch.no_grad()
def sdae_k_sweep_mse(model: nn.Module,
                     loader,
                     ks: List[int],
                     noise_factor: float,
                     device: torch.device,
                     criterion = nn.MSELoss()) -> Dict[int, float]:
    """
    Compute test MSE for each k re-feeds on the *same* test set, denoising mode.
    Returns dict {k: mse}.
    """
    results = {}
    for k in ks:
        mse_k = evaluate(model, loader, denoising=True, re_feeds=k,
                         noise_factor=noise_factor, device=device, criterion=criterion)
        results[k] = mse_k
    return results

# Baseline reference (single numbers)
dae_mse  = evaluate(dae, test_loader, denoising=True,  re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)
ae_mse   = evaluate(ae,  test_loader, denoising=False, re_feeds=1, noise_factor=noise_factor, device=device, criterion=criterion)

ks = list(range(1, 101))
SGD_momentum_Nesterov = sdae_k_sweep_mse(sdae, test_loader, ks, noise_factor, device, criterion)

# ----------------------------
# Plot
# ----------------------------
plt.figure(figsize=(9, 4.5))
plt.plot(ks, [SGD_momentum_Nesterov[k] for k in ks], marker='o', linestyle='-', label='SDAE (k re-feeds)')
plt.axhline(dae_mse, linestyle='--', label=f'DAE (1 pass): {dae_mse:.4f}')
plt.axhline(ae_mse,  linestyle=':',  label=f'AE (clean→clean): {ae_mse:.4f}')
plt.xlabel('Number of Re-feeds (k)')
plt.ylabel('Test Reconstruction MSE vs. Clean')
plt.title('SDAE Re-feed Depth vs. Test MSE (MNIST)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

# ----------------------------
# Optional: quick sample visualization of denoising
# ----------------------------
@torch.no_grad()
def visualize_examples(model_sdae, model_dae, model_ae, loader, k_sdae=3, noise_factor=0.81, num_show=8):
    imgs, _ = next(iter(loader))
    imgs = imgs[:num_show].to(device)
    noisy = add_noise(imgs, noise_factor)

    deno_sdae = forward_k(model_sdae, noisy, k_sdae)
    deno_dae  = model_dae(noisy)
    recon_ae  = model_ae(imgs)

    # grid: clean | noisy | SDAE | DAE | AE (for reference)
    grid = torch.cat([imgs.cpu(), noisy.cpu(), deno_sdae.cpu(), deno_dae.cpu(), recon_ae.cpu()], dim=0)
    grid = make_grid(grid, nrow=num_show, pad_value=1.0)
    plt.figure(figsize=(num_show * 1.2, 6))
    plt.imshow(grid.permute(1, 2, 0))
    plt.axis("off")
    plt.title(f"SDAE k={k_sdae} | DAE 1-pass | AE 1-pass")
    plt.show()

# Uncomment to visualize examples after training:
# visualize_examples(sdae, dae, ae, test_loader, k_sdae=stack_k_eval, noise_factor=noise_factor, num_show=8)
```
